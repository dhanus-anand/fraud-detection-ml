# 01 - Exploratory Data Analysis

**Dataset:** Kaggle Credit Card Fraud Detection  
**Size:** ~284,807 transactions, 492 fraud cases (0.172%)  

## Goals
1. Understand the dataset structure
2. Quantify class imbalance
3. Explore feature distributions
4. Identify correlations with fraud label
5. Inform feature engineering strategy

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', None)

DATA_PATH = Path('../data/raw/creditcard.csv')
```

## 1. Load Data

```python
df = pd.read_csv(DATA_PATH)
print(f'Shape: {df.shape}')
df.head()
```

## 2. Class Imbalance Analysis

```python
class_counts = df['Class'].value_counts()
fraud_rate = class_counts[1] / len(df) * 100

print(f'Total transactions: {len(df):,}')
print(f'Legitimate: {class_counts[0]:,} ({100 - fraud_rate:.3f}%)')
print(f'Fraud: {class_counts[1]:,} ({fraud_rate:.3f}%)')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
class_counts.plot(kind='bar', ax=axes[0], color=['steelblue', 'crimson'])
axes[0].set_title('Class Distribution (Count)')
axes[0].set_xticklabels(['Legitimate', 'Fraud'], rotation=0)

axes[1].pie([class_counts[0], class_counts[1]], labels=['Legitimate', 'Fraud'],
            autopct='%1.3f%%', colors=['steelblue', 'crimson'])
axes[1].set_title('Class Distribution (Proportion)')
plt.tight_layout()
plt.show()
```
