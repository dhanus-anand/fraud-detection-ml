## 2. Threshold Optimization

```python
def expected_cost_at_threshold(y_true, y_proba, threshold, fraud_cost, decline_cost):
    """Calculate expected business cost for a given classification threshold."""
    y_pred = (y_proba >= threshold).astype(int)
    
    # False negatives: actual fraud predicted as legitimate
    fn = ((y_pred == 0) & (y_true == 1)).sum()
    # False positives: legitimate predicted as fraud (declined)
    fp = ((y_pred == 1) & (y_true == 0)).sum()
    
    total_cost = (fn * fraud_cost) + (fp * decline_cost)
    cost_per_1000 = total_cost / len(y_true) * 1000
    return cost_per_1000

# TODO: Sweep thresholds from 0.05 to 0.95
# TODO: Plot cost vs threshold curve
# TODO: Mark optimal threshold
# TODO: Show precision/recall at optimal threshold
```
